```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import kurtosis
from sklearn.ensemble import IsolationForest

np.random.seed(0)
t = np.linspace(0, 1, 2000)

def make_window(fault=0):
    s = np.sin(2*np.pi*50*t) + 0.5*np.sin(2*np.pi*120*t) + 0.2*np.random.randn(len(t))
    if fault > 0:
        idx = np.random.choice(len(t), 10, replace=False)
        s[idx] += fault * np.random.choice([-1, 1], 10) * 3
    return s

# 150 نافذة سليمة ثم 50 نافذة يتزايد فيها العطل
windows = [make_window(0) for _ in range(150)] + \
          [make_window(i/50) for i in range(1, 51)]

def features(s):
    rms = np.sqrt(np.mean(s**2))
    peak = np.max(np.abs(s))
    return [rms, kurtosis(s), peak, peak/rms]

X = np.array([features(w) for w in windows])

# تدريب النموذج على الحالة السليمة فقط
model = IsolationForest(random_state=0).fit(X[:100])
score = -model.score_samples(X)          # أعلى = أكثر شذوذاً
limit = np.percentile(score[:100], 99)   # حد الإنذار

plt.plot(score, label="Anomaly score")
plt.axhline(limit, color="orange", linestyle="--", label="Alarm limit")
plt.axvline(150, color="red", linestyle=":", label="Fault starts")
plt.legend()
plt.show()
above = score > limit
alarm = next((i for i in range(len(above)-2)
              if above[i] and above[i+1] and above[i+2]), None)

print("بداية العطل الفعلية: 150")
print("أول إنذار:", alarm)
print("التأخر في الكشف:", alarm - 150, "قياس")
print("إنذارات كاذبة قبل العطل:", int(above[:150].sum()))
for pct in [90, 95, 99]:
    lim = np.percentile(score[:100], pct)
    ab = score > lim
    for n in [1, 2, 3]:
        a = next((i for i in range(100, len(ab)-n+1) if ab[i:i+n].all()), None)
        status = "إنذار كاذب" if a is not None and a < 150 else "صحيح"
        print(f"مئين {pct} | متتالية {n} | أول إنذار: {a} | {status}")
        k = 10  # متوسط متحرك على 10 قياسات
smooth = np.convolve(score, np.ones(k)/k, mode="valid")

lim = np.percentile(smooth[:100-k+1], 99)
ab = smooth > lim

first = next((i+k-1 for i in range(100-k+1, len(ab)) if ab[i]), None)
false_alarms = sum(1 for i in range(len(ab)) if ab[i] and i+k-1 < 150)

print("أول إنذار:", first)
print("إنذارات كاذبة قبل العطل:", false_alarms)
for k in [3, 5, 10]:
    model = IsolationForest(random_state=0).fit(X[:100])
    score = -model.score_samples(X)
    smooth = np.convolve(score, np.ones(k)/k, mode="valid")
    pad = np.concatenate([np.full(k-1, np.nan), smooth])

    lim = np.nanmax(pad[100:150])   # الحد من بيانات سليمة غير مرئية
    first = next((i for i in range(150, len(pad)) if pad[i] > lim), None)

    print(f"k={k} | الحد: {lim:.3f} | أول إنذار: {first} | التأخر: {first-150 if first else None}")
```
